```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 100 linearly spaced numbers for the effort level (in hr)
e = np.linspace(0,100,num=100)
# effort cost function
effort_cost=1000+e**2
# firm's benefit of effort
benefit=100*e
# firms profit
profits= benefit-effort_cost
```

```python
# setting the axes at the centre
fig = plt.figure()
ax=plt.axes(xlim=(0, 100),ylim=(0, 10000))

# plot the function
plt.plot(e,effort_cost, 'r', label='effort cost')
plt.plot(e,benefit, 'b', label='firm benefits of effort')
plt.plot(e,profits, 'g', label='firm profits')
plt.axvline(x=50)


# show the plot
plt.xlabel("employee's effort (in hr)")
plt.ylabel("Benefits and cost ($)")
plt.legend(loc='upper left')
plt.show()
```
